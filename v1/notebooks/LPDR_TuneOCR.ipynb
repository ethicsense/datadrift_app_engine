```python
import time

data_path = '../datasets/KLPOCR_dataset/Training/images/'
file_names = os.listdir(data_path)
cnt = 0

for index, file_name in enumerate(file_names):
    # 파일 확장자 분리
    name, ext = os.path.splitext(file_name)
    
    # "-"가 있는 경우, "-" 앞의 문자열만 사용
    if '-' in name:
        label = name.split('-')[0]
    elif '_' in name:
        label = name.split('_')[0]
    else:
        label = name
    
    # 새로운 파일명 생성
    new_file_name = f"{label}_{index}{ext}"
    cnt += 1
    
    # 파일명 변경
    os.rename(os.path.join(data_path, file_name), os.path.join(data_path, new_file_name))
    print(f'Renamed: {file_name} -> {new_file_name}')
    if cnt % 20 == 0:
        print(f"{cnt}개 파일 처리 완료")
        time.sleep(1)
```

```python
import lmdb
import cv2
import numpy as np
import matplotlib.pyplot as plt
import random

def read_lmdb(lmdb_path):
    # Open the LMDB environment
    env = lmdb.open(lmdb_path, readonly=True)
    
    with env.begin() as txn:
        # Get the number of samples
        num_samples = int(txn.get('num-samples'.encode()).decode())
        print(f"Number of samples: {num_samples}")

        # 랜덤하게 20개의 샘플 인덱스 선택
        sample_indices = random.sample(range(1, num_samples + 1), 20)
        
        for i in sample_indices:
            # Read image
            image_key = f'image-{i:09d}'.encode()
            image_bin = txn.get(image_key)
            if image_bin is not None:
                image_buf = np.frombuffer(image_bin, dtype=np.uint8)
                image = cv2.imdecode(image_buf, cv2.IMREAD_COLOR)
                
                # Display image using matplotlib
                plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
                plt.axis('off')  # Hide axes
                plt.title(f"Sample {i}")
                plt.show()

            # Read label
            label_key = f'label-{i:09d}'.encode()
            label = txn.get(label_key)
            if label is not None:
                print(f"Label: {label.decode('utf-8')}")

    cv2.destroyAllWindows()

# Example usage
lmdb_path = '../datasets/KLPOCR_dataset/lmdb/train/'
read_lmdb(lmdb_path)
```

```python
import lmdb
import matplotlib.pyplot as plt
import numpy as np
import cv2
import os
import unicodedata

def find_label_in_lmdb(base_path, target_label):
    # Define the subdirectories to search
    subdirs = ['train', 'val']
    
    for subdir in subdirs:
        lmdb_path = os.path.join(base_path, subdir)
        print(f"Searching in {lmdb_path}...")
        
        # Check if the LMDB path exists
        if not os.path.exists(lmdb_path):
            print(f"LMDB path {lmdb_path} does not exist.")
            continue
        
        # Open the LMDB environment
        env = lmdb.open(lmdb_path, readonly=True)
        
        with env.begin() as txn:
            # Get the number of samples
            num_samples = txn.get('num-samples'.encode())
            if num_samples is None:
                print(f"No samples found in {lmdb_path}.")
                continue
            num_samples = int(num_samples.decode())
            print(f"Number of samples in {subdir}: {num_samples}")
            
            for i in range(1, num_samples + 1):
                # Read label
                label_key = f'label-{i:09d}'.encode()
                label = txn.get(label_key)
                if label is not None:
                    decoded_label = label.decode('utf-8').strip()
                    normalized_label = unicodedata.normalize('NFC', decoded_label)
                    # print(f"Checking label: {normalized_label}")  # 디버깅 출력
                    if normalized_label == target_label.strip():
                        print(f"Found label '{target_label}' in {subdir} at index {i}")
                        
                        # Read and display the corresponding image
                        image_key = f'image-{i:09d}'.encode()
                        image_bin = txn.get(image_key)
                        if image_bin is not None:
                            print(f"Image key: {image_key} found")  # 디버깅 출력
                            image_buf = np.frombuffer(image_bin, dtype=np.uint8)
                            image = cv2.imdecode(image_buf, cv2.IMREAD_COLOR)
                            if image is not None:
                                # Display image using matplotlib
                                plt.imshow(cv2.cvtColor(image, cv2.COLOR_BGR2RGB))
                                plt.axis('off')  # Hide axes
                                plt.title(f"Label: {normalized_label} in {subdir}")
                                plt.show()
                            else:
                                print("Failed to decode image")  # 디버깅 출력
                        else:
                            print(f"Image key: {image_key} not found")  # 디버깅 출력
                        return  # Exit after finding the first match
        print(f"Label '{target_label}' not found in {subdir}.")

# Example usage
base_path = '../datasets/KLPOCR_dataset/lmdb'
target_label = "88너1094"
find_label_in_lmdb(base_path, target_label)
```
